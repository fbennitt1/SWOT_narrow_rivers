```python
import os

import contextily as ctx
import geopandas as gpd
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import xarray as xr

import hvplot
import hvplot.xarray

import rioxarray

import matplotlib.colors as colors
```

```python
# From https://podaac.github.io/tutorials/notebooks/datasets/SWOTHR_localmachine.html

# function to get interferogram from slc images
def interfere(slc):
    # convert to complex array
    slc_plus_y = slc.slc_plus_y[:,:,0] + 1j * slc.slc_plus_y[:,:,1]
    slc_minus_y = slc.slc_minus_y[:,:,0] + 1j * slc.slc_minus_y[:,:,1]
    # get interferogram quantities
    ifgram = slc_plus_y * np.conj(slc_minus_y)
    mag = np.abs(ifgram)
    phase = np.arctan2(np.imag(ifgram), np.real(ifgram))
    coh = (slc_plus_y * np.conj(slc_minus_y)).real / np.sqrt(np.abs(slc_plus_y)**2 * np.abs(slc_minus_y)**2).real
    coh_pow = 0.5 * np.abs(slc_plus_y + slc_minus_y)**2
    return ifgram, np.real(np.abs(slc_plus_y)), np.real(np.abs(slc_minus_y)), mag, phase, coh, coh_pow
```

### Fort River

```python
slc_path = os.path.join(data_path, 'SWOT_L1B_HR_SLC_032_298_080L_*')
slc = xr.open_mfdataset(slc_path, group='slc', engine='h5netcdf')

# ignore the divide-by-zero warning in log10
import warnings
warnings.filterwarnings("ignore")

ifgram, power_plus_y, power_minus_y, mag, phase, coh, coh_pow = interfere(slc)
```
