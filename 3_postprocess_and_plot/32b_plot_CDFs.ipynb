## Stats

```python
data_dict = {
    # ' > 80': df_80['wse_diff'],
    ' > 50': df_50['wse_diff'],
    '50-60': df_50_60['wse_diff'],
    '40-50': df_40_50['wse_diff'],
    '30-40': df_30_40['wse_diff'],
    '20-30': df_20_30['wse_diff'],
    '10-20': df_10_20['wse_diff'],
    # '04-10': df_04_10['wse_diff_abs']
}

group_names = list(data_dict.keys())
```

```python
from scipy.stats import ks_2samp
from statsmodels.stats.multitest import multipletests
import pandas as pd
from itertools import combinations

# 1. Get all unique pairs of your groups
group_names = list(data_dict.keys())
pairs = list(combinations(group_names, 2))

# 2. Run KS test for each pair
ks_list = []
for g1, g2 in pairs:
    # Dropna is crucial for KS test stability
    d1 = data_dict[g1].dropna()
    d2 = data_dict[g2].dropna()
    
    stat, p_val = ks_2samp(d1, d2)
    ks_list.append([g1, g2, stat, p_val])

# 3. Create DataFrame and apply Holm correction
ks_df = pd.DataFrame(ks_list, columns=['Group 1', 'Group 2', 'KS Stat', 'p-value'])
ks_df['p-adj'] = multipletests(ks_df['p-value'], method='holm')[1]
ks_df['Significant'] = ks_df['p-adj'] < 0.05

print(ks_df.sort_values('p-adj'))
```

```python
results_expanded = []
for g1, g2, stat, p_val in ks_list:
    results_expanded.append({'Group A': g1, 'Group B': g2, 'p_adj': p_val}) # using raw p for now
    results_expanded.append({'Group A': g2, 'Group B': g1, 'p_adj': p_val})

# Add the diagonal (1.0)
for g in group_names:
    results_expanded.append({'Group A': g, 'Group B': g, 'p_adj': 1.0})

ks_matrix_df = pd.DataFrame(results_expanded)

# 2. Pivot into the final matrix
ks_posthoc_matrix = ks_matrix_df.pivot(index='Group A', columns='Group B', values='p_adj')

# 3. Apply the Multi-test correction (Holm) to the entire matrix 
# (Excluding the diagonal 1.0s to stay accurate)
p_values = ks_posthoc_matrix.values
mask = ~np.eye(p_values.shape[0], dtype=bool)
p_values[mask] = multipletests(p_values[mask], method='holm')[1]

ks_posthoc_matrix = pd.DataFrame(p_values, index=ks_posthoc_matrix.index, columns=ks_posthoc_matrix.columns)

# Reorder to match your original data_dict order
ks_posthoc_matrix = ks_posthoc_matrix.reindex(index=group_names, columns=group_names)

print(ks_posthoc_matrix)
```

```python
import matplotlib.patches as mpatches
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# 1. Setup categorical data (1 for Sig, 0 for Not Sig)
sig_matrix = (ks_posthoc_matrix < 0.05).astype(int)

# Use k=0 to match the mask style of the second plot (hides diagonal and upper triangle)
mask = np.triu(np.ones_like(sig_matrix, dtype=bool), k=0)

# 2. Define colors and palette from the second plot
colors = ["lemonchiffon", "royalblue"]  # [Not Sig, Significant]
cmap = sns.color_palette(colors)

# Setup figure
plt.figure(figsize=(10, 8))

# 3. Plot with second plot's aesthetics (square cells, clean lines, bold text)
ax = sns.heatmap(
    sig_matrix,
    mask=mask,
    annot=ks_posthoc_matrix,  # Keep original p-values as text
    fmt=".3f",
    cmap=cmap,
    cbar=False,  # Turn off colorbar
    square=True,  # Keeps cells perfectly square
    annot_kws={
        "size": 10,
        "weight": "bold",
    },  # Bold annotations for contrast
)

# Optional labels to match the axis label style of the second plot
ax.set_xlabel("X Label Here", fontsize=14)
ax.set_ylabel("Y Label Here", fontsize=14)
ax.tick_params(axis="x", labelsize=12)
ax.tick_params(axis="y", labelsize=12)

# 4. Create the formatted legend matching the second plot's font properties
not_sig_patch = mpatches.Patch(
    color=colors[0], label="Not Significant (p ≥ 0.05)"
)
sig_patch = mpatches.Patch(color=colors[1], label="Significant (p < 0.05)")

plt.legend(
    handles=[not_sig_patch, sig_patch],
    loc="upper center",
    bbox_to_anchor=(0.5, 1.1),  # Centers it cleanly above the single plot
    ncol=2,
    frameon=False,
    prop={"size": 12, "weight": "semibold"},
)

plt.xlabel('River width (m)')
plt.ylabel('River width (m)')

# plt.tight_layout()
# plt.show()

save_dir = '/nas/cee-water/cjgleason/fiona/for_manuscripts/26_narrow_rivers/results_figures/'
plt.savefig(f'{save_dir}ks_test_bias.png', dpi=300, bbox_inches='tight')
```

## Plots

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.lines import Line2D

def plot_river_cdf(all_dfs, labels, plot_indices=None,
                   xlim_max=5, title='relative_cdf'):
    if plot_indices is None:
        plot_indices = range(len(all_dfs))

    plt.figure(figsize=(19, 9))


    colors = [
    '#FF7F0E',  # 2. Classic tab10 Orange
    '#1F77B4',  # 1. Classic tab10 Blue
    '#8DAE91',  # 3. YOUR ORIGINAL BASE GREEN (Replaces standard green)
    'blue',   # 10. Deep Indigo Blue
    'olive',
    '#004D40',  # 8. Deep Pine Green (High contrast substitute for red/brown)    
    ]


    # --- Thresholds & Zero Line ---
    plt.axhline(y=0.68, color='dimgrey', linestyle='--', alpha=0.7)
    plt.axhline(y=0.5, color='dimgray', linestyle='--', alpha=0.7)
    plt.axvline(x=0, color='black', linewidth=1.5, alpha=0.8) # Darker tick at 0
    
    signed_handles = []
    absolute_handles = []
    dashes = [
        'solid', #placeholder
        ':',
        '--',
        ':',
        '--',
        ':'
    ]
    
    for i in plot_indices:
        df = all_dfs[i]
        label = labels[i]
        color = colors[i]
        dash_style = dashes[i]
        
        # 1. Plot Signed (Solid)
        data_signed = np.sort(df['wse_diff'].dropna())
        y_signed = np.arange(len(data_signed)) / float(len(data_signed) - 1)
        line_s, = plt.plot(data_signed, y_signed, linewidth=3, 
                           color=color, linestyle='-')
        # Store handle with specific label for left column
        line_s.set_label(f'{label} (signed)')
        signed_handles.append(line_s)
        
        # 2. Plot Absolute (Dashed)
        data_abs = np.sort(df['wse_diff_abs'].dropna())
        y_abs = np.arange(len(data_abs)) / float(len(data_abs) - 1)
        line_a, = plt.plot(data_abs, y_abs, linewidth=4, 
                           color=color, linestyle=dash_style
                           # ='--'
                          )
        # Store handle with specific label for right column
        line_a.set_label(f'{label} (unsigned)    n={len(data_abs)}') 
        absolute_handles.append(line_a)
        
    # --- Custom Legend Logic ---
    # Concatenate so signed come first, then absolute
    # Matplotlib fills columns first when ncol is used
    all_handles = signed_handles + absolute_handles
    
    plt.legend(handles=all_handles, title='River width (m)', title_fontsize=18, fontsize=16,
               loc='upper left', ncol=2, columnspacing=1.0)

    # Styling and Ticks
    plt.text(-xlim_max * 1.075, 0.68, '0.68', verticalalignment='center',
             color='dimgray', fontsize=16, fontweight='semibold')
    plt.text(-xlim_max * 1.055, 0.5, '0.5', verticalalignment='center',
             color='dimgray', fontsize=16, fontweight='semibold')
    
    plt.xlim(-xlim_max, xlim_max)
    plt.ylim(0, 1.001)
    ax = plt.gca() 

    if xlim_max == 1:
        plt.xticks(np.arange(-xlim_max, xlim_max + 0.1, 0.2))
        ax.xaxis.set_minor_locator(plt.MultipleLocator(0.1))
    else:
        plt.xticks(np.arange(-xlim_max, xlim_max + 0.1, 0.5))
        ax.xaxis.set_minor_locator(plt.MultipleLocator(0.25))

    plt.grid(which='minor', axis='x', linestyle=':', alpha=0.4)
    plt.grid(which='major', axis='both', linestyle='--', alpha=0.6)
    plt.xlabel('Relative (median-normalized) WSE difference (m)', fontsize=18, fontweight='bold')
    plt.ylabel('Proportion of observations', fontsize=18, fontweight='bold')
    plt.xticks(fontsize=16, fontweight='semibold')
    plt.yticks(fontsize=16, fontweight='semibold')

    plt.savefig(f'{save_dir}relative_wse_{title}.png', dpi=300, bbox_inches='tight')
```
